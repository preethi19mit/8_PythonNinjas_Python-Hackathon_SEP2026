```python
# code to check for duplicate rows in the responsiveness dataset
import pandas as pd

# Load the responsiveness data
df = pd.read_csv('responsivenes.csv')

# Check if there are any duplicates
print(f"Has duplicates: {df.duplicated().any()}")

# Count total duplicate rows
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# View all duplicate rows
duplicate_rows = df[df.duplicated(keep=False)]
print(f"\nDuplicate rows:\n{duplicate_rows}")

# Count unique rows
print(f"\nTotal rows: {len(df)}")
print(f"Unique rows: {len(df.drop_duplicates())}")

# Remove duplicates
df_cleaned = df.drop_duplicates()
print(f"Rows after removing duplicates: {len(df_cleaned)}")

# Save cleaned version (optional)
df_cleaned.to_csv('responsivenes_cleaned.csv', index=False)
```
